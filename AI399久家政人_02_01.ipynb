```python
import google.generativeai as genai
# APIキーをcolabから取得してAIへ設定する
from google.colab import userdata
GOOLE_API_KEY = userdata.get('GOOGLE_API_KEY')
genai.configure( api_key=GOOLE_API_KEY)
# APIキーが設定済みのモデルインスタンスを生成
model = genai.GenerativeModel("gemini-3.6-flash")
```

```python
response = model.generate_content("こんにちは！　私は政人と言います！")
print( response.text )
```

## 履歴を送信する

```python
chat = model.start_chat(
    history=[
        {"role":"user", "parts":["こんにちは！私は政人と言います！"]},
        {"role":"model", "parts":["こんにちは政人さん初めまして！　何かお手伝いしましょうか"]}
    ]
)
response = chat.send_message("私の名前がわかりますか？")
print(response.text)
```

### ストーリム応答を試す

```python
# modelの生成までは上のものを流用
responce = model.generate_content(
    "こんにちは、私は久家と言います。一行100文字4行で返事してください",
    stream=True
)
# 応答を順次表示する
for chunk in responce:
  print( chunk.text, end="", flush=True )
```

## JSON形式で応答を受け取る

```python
# modelの生成までは上のものを流用する
response = model.generate_content(
    "昔々あるところにおじいさんとおばあさんがいました。登場人物を次のJSON形式で出力してください。{'people':['aaa', 'bbb']}",
    generation_config={"response_mime_type":"application/json"}

)
print(response.text)
```

## AIに画像を分析させる

```python
# 必要なライブラリの追加 import
import requests
from PIL import Image
from io import BytesIO

image_url = "https://raw.githubusercontent.com/yoshidashingo/langchain-book/main/assets/cover.jpg"
response_img = requests.get(image_url)
img = Image.open(BytesIO(response_img.content))

# modelの生成は上の部分を流用
response = model.generate_content([
    "この画像を説明してください。",
    img
])
print(response.text)

```
